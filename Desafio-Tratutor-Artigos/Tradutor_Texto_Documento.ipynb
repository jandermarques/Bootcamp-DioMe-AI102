```python
# Imports and parameters
import requests
from docx import Document
import os

SUBSCRIPTION_KEY = "9b9e6b3e4b3f4b3e4b3f4b3e4b3f4b3e"
ENDPOINT = "https://api.cognitive.microsofttranslator.com"
LOCATION = "eastus2"
TARGET_LANGUAGE = "pt-br"  # Portuguese (Brazil)    
```

```python
def translator_text(text, TARGET_LANGUAGE):
    path = '/translate'
    constructed_url = ENDPOINT + path

    headers = {
        'Ocp-Apim-Subscription-Key': SUBSCRIPTION_KEY,
        'Ocp-Apim-Subscription-Region': LOCATION,
        'Content-type': 'application/json',
        'X-ClientTraceId': str(os.urandom(16))
    }
    body = [{
        'text': text
    }]
    params = {
        'api-version': '3.0', 
        'from': 'en', # English
        'to': TARGET_LANGUAGE
    }
    
    request = requests.post(constructed_url, params = params, headers = headers, json = body)
    response = request.json()
    return response[0]['translations'][0]['text']
```

```python
translator_text("For I am the Lord your God who takes hold of your right hand and says to you: Do not fear; I will help you.’", TARGET_LANGUAGE)
```

```python
def translate_document(path, TARGET_LANGUAGE):
    doc = Document(path)

    full_text = []
    for paragraph in doc.paragraphs:
        if paragraph.text.strip():  # Check if the paragraph is not empty
            translated_text = translator_text(paragraph.text, TARGET_LANGUAGE)
            full_text(translated_text)

    translates_doc = Document()
    for line in full_text:
        translates_doc.add_paragraph(line)

    path_translated = path.replace(".docx", f"_{TARGET_LANGUAGE}.docx")
    translates_doc.save(path_translated)

    return path_translated
```

```python
input_file = "/content/ARTIGO.docx"
translate_document(input_file, TARGET_LANGUAGE)
```
